## 1. Method choice and why

*Which method from the toolkit, and why it fits your lane.*


I am working on **Lane 2: Refresh / Content Opportunity Scoring**, where the main objective is to rank content pages so that a human reviewer can decide which pages should be reviewed first.

I selected a **Random Forest classifier** because the task contains multiple observable signals that may interact with one another, including impressions, clicks, sessions, CTR, average position, content age, freshness, word count, and engagement. Random Forest can capture non-linear relationships between these features while still being relatively easier to inspect than more complex models.

The target variable is the defined proxy **`is_declining_label`**, where `1` indicates that the observed `trend_direction` is `"down"`. This label represents an observed decline in the available data. It should not be treated as a guarantee of future performance or as evidence that refreshing a page will cause an improvement.

The model will be evaluated as a **ranking system using Precision@50**, since the practical use case is to select a limited number of pages for initial human review. The model is considered useful only if it produces a better-quality review queue than the **Week-4 baseline**.


```python
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier

print("Libraries loaded successfully.")
print("Method: Random Forest")
print("Target: is_declining_label")
print("Metric: Precision@50")
```

## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*


I use a **client-grouped split** to ensure that pages from the same client are not present in both the training and test datasets.

This provides a more honest evaluation for this task because pages belonging to the same client may share similar characteristics. If a client were included in both datasets, the model could learn client-specific patterns instead of showing whether it can generalize to clients it has not seen during training.

I assign **80% of the clients to the training set and 20% to the test set**. The test clients remain completely isolated from the model during training.

The split is created before fitting the model, and the test set is used only for the final comparison against the **Week-4 baseline**.

```python
from pathlib import Path

repo_path = Path("/content/Anshika-FlyRank-ML")

if not repo_path.exists():
    !git clone -q https://github.com/Anshikaag-28/Anshika-FlyRank-ML.git /content/Anshika-FlyRank-ML

%cd /content/Anshika-FlyRank-ML

data_path = Path("data/raw/content_refresh_anonymized.csv")

df = pd.read_csv(data_path)

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
# Create the proxy target defined in the earlier task framing

df["is_declining_label"] = (
    df["trend_direction"]
    .astype(str)
    .str.lower()
    .eq("down")
    .astype(int)
)

print("Target created successfully.")
print(df["is_declining_label"].value_counts())
```

```python
feature_candidates = [
    "impressions_90d",
    "sessions_90d",
    "content_age_days",
    "freshness_days",
    "ctr",
    "avg_position",
    "word_count",
    "engagement_rate"
]

features = [c for c in feature_candidates if c in df.columns]

print("Features being used:")
print(features)
```

```python
from sklearn.model_selection import train_test_split

# Check that the correct client identifier exists
assert "client_id" in df.columns

# Keep rows with required information
model_df = df.dropna(
    subset=features + ["is_declining_label", "client_id"]
).copy()

# Get unique clients
clients = model_df["client_id"].unique()

# Split CLIENTS, not individual rows
from sklearn.model_selection import train_test_split

train_clients, test_clients = train_test_split(
    clients,
    test_size=0.20,
    random_state=42
)

# Create train and test datasets
train_df = model_df[
    model_df["client_id"].isin(train_clients)
].copy()

test_df = model_df[
    model_df["client_id"].isin(test_clients)
].copy()

print("Total clients:", len(clients))
print("Training clients:", len(train_clients))
print("Test clients:", len(test_clients))

print("Training rows:", len(train_df))
print("Test rows:", len(test_df))
```

```python
# Check that no client appears in both datasets

overlap = set(train_df["client_id"]).intersection(
    set(test_df["client_id"])
)

print("Client overlap:", len(overlap))

assert len(overlap) == 0

print("PASS: No client appears in both train and test.")
```

## 3. Train + compare vs my baseline

*Same data, same metric, same split as your Week-4 baseline. Show the table.*


I train the **Random Forest** using only the observable features defined earlier. The model then assigns a probability to each page indicating how likely it is to belong to the positive proxy group.

For evaluation, I sort the test pages according to their predicted probability and calculate **Precision@50** based on the top 50 pages.

I compare the Random Forest results with the **Week-4 baseline using the same test population and the same Precision@50 definition**. This ensures that both approaches are evaluated fairly on exactly the same set of pages.

The goal is not to assume that the more complex model is automatically better. The Random Forest should only be considered useful if it produces better ranking quality than the transparent baseline.


```python
# Prepare train and test data

X_train = train_df[features].copy()
y_train = train_df["is_declining_label"].copy()

X_test = test_df[features].copy()
y_test = test_df["is_declining_label"].copy()

# Fill missing numeric values using training medians
train_medians = X_train.median(numeric_only=True)

X_train = X_train.fillna(train_medians)
X_test = X_test.fillna(train_medians)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
# Train Random Forest

rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=12,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

rf_model.fit(X_train, y_train)

print("Random Forest training completed.")
```

```python
def precision_at_k(y_true, scores, k=50):
    """
    Precision@K:
    Number of positive pages among the top K ranked pages / K
    """
    y_true = np.asarray(y_true)
    scores = np.asarray(scores)

    k = min(k, len(y_true))

    ranking = np.argsort(scores)[::-1][:k]

    return y_true[ranking].mean()


# Model probabilities
model_scores = rf_model.predict_proba(X_test)[:, 1]

model_precision_50 = precision_at_k(
    y_test,
    model_scores,
    k=50
)

print(f"Random Forest Precision@50: {model_precision_50:.3f}")
```

```python
# Recreate the Week-4 transparent baseline on the test set

baseline_test = test_df.copy()

# Create position buckets
baseline_test["position_bucket"] = pd.cut(
    baseline_test["avg_position"],
    bins=[0, 3, 10, 20, 1000],
    labels=["Top 3", "Page 1", "Page 2", "Deep"],
    include_lowest=True
)

# Calculate expected CTR from the TRAINING data only.
# This avoids using test information to construct the benchmark.

baseline_train = train_df.copy()

baseline_train["position_bucket"] = pd.cut(
    baseline_train["avg_position"],
    bins=[0, 3, 10, 20, 1000],
    labels=["Top 3", "Page 1", "Page 2", "Deep"],
    include_lowest=True
)

expected_ctr = (
    baseline_train
    .groupby("position_bucket", observed=True)["ctr"]
    .median()
)

baseline_test["expected_ctr"] = (
    baseline_test["position_bucket"]
    .map(expected_ctr)
)

baseline_test["expected_ctr"] = pd.to_numeric(
    baseline_test["expected_ctr"],
    errors="coerce"
)

baseline_test["ctr"] = pd.to_numeric(
    baseline_test["ctr"],
    errors="coerce"
)

baseline_test["impressions_90d"] = pd.to_numeric(
    baseline_test["impressions_90d"],
    errors="coerce"
)

baseline_test["ctr_gap"] = (
    baseline_test["expected_ctr"] -
    baseline_test["ctr"]
)

baseline_test["baseline_score"] = (
    baseline_test["ctr_gap"].clip(lower=0) *
    np.log1p(baseline_test["impressions_90d"])
)

baseline_test["baseline_score"] = (
    baseline_test["baseline_score"]
    .fillna(0)
)

baseline_precision_50 = precision_at_k(
    baseline_test["is_declining_label"],
    baseline_test["baseline_score"],
    k=50
)

print(f"Week-4 baseline Precision@50: {baseline_precision_50:.3f}")
```

```python
comparison = pd.DataFrame({
    "Method": [
        "Week-4 baseline",
        "Random Forest"
    ],
    "Precision@50": [
        baseline_precision_50,
        model_precision_50
    ]
})

comparison
```

## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

The main errors in the ranked review queue are **false positives and false negatives**. A false positive occurs when the model ranks a page highly even though its proxy label is not positive. A false negative occurs when a page has a positive proxy label but is ranked too low by the model.

I review the top-ranked test pages and compare their observed signals with their proxy labels. This helps identify situations where factors such as high impressions, CTR, average position, or sessions cause the model to rank a page highly even when the proxy label does not support that ranking.

Feature importance is used for **interpretation rather than causation**. If a feature has high importance, it means the model relied on that feature when generating its predictions; it does not mean that the feature itself causes a page to decline.

Therefore, the model should be treated as a **decision-support ranking tool**, rather than as a guarantee that a particular page needs a refresh or that refreshing it will improve its performance.


```python
# Create test predictions

error_df = test_df[
    ["client_id", "content_id", "is_declining_label"] +
    features
].copy()

error_df["model_probability"] = model_scores

error_df["predicted_positive"] = (
    error_df["model_probability"] >= 0.50
).astype(int)

# False positives
false_positives = error_df[
    (error_df["predicted_positive"] == 1) &
    (error_df["is_declining_label"] == 0)
].copy()

# False negatives
false_negatives = error_df[
    (error_df["predicted_positive"] == 0) &
    (error_df["is_declining_label"] == 1)
].copy()

print("False positives:", len(false_positives))
print("False negatives:", len(false_negatives))
```

```python
top_20 = (
    error_df
    .sort_values("model_probability", ascending=False)
    .head(20)
)

display(
    top_20[
        [
            "client_id",
            "content_id",
            "model_probability",
            "is_declining_label"
        ] + features
    ]
)
```

```python
importance = pd.DataFrame({
    "feature": features,
    "importance": rf_model.feature_importances_
}).sort_values(
    "importance",
    ascending=False
)

display(importance)
```

```python
print("Most important features according to the Random Forest:")

for _, row in importance.head(5).iterrows():
    print(
        f"- {row['feature']}: "
        f"{row['importance']:.3f}"
    )
```

```python
forbidden_features = {
    "trend_direction",
    "trend_pct",
    "is_declining_label",
    "priority_score",
    "action_type",
    "health_score"
}

leaked_features = set(features).intersection(
    forbidden_features
)

print("Forbidden features found:", leaked_features)

assert len(leaked_features) == 0

print("PASS: No target/product-decision fields are used as model features.")
```

```python
assert set(train_df["client_id"]).isdisjoint(
    set(test_df["client_id"])
)

print("PASS: Client-grouped split is leakage-safe.")
```

### Final interpretation

On the held-out test clients, the **Random Forest achieved a Precision@50 of 0.84**, while the **Week-4 baseline achieved 0.70**. This means the model produced better top-50 ranking quality on the selected test split.

This result comes from the selected anonymized dataset, so it should be considered **directional evidence rather than a general benchmark**. The Random Forest uses multiple observable signals together instead of depending on a single fixed rule.

However, the error analysis shows that a high model score does not necessarily mean that a page actually requires a refresh. Therefore, the model output should be used as **decision support for human review**, rather than being treated as an automatic refresh action.
