```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Dataset shape:", df.shape)

# Basic statistics relevant to CTR / Engagement Opportunity Scoring
total_pages = df["content_id"].nunique()

# Pages with meaningful search visibility
visible_pages = df[df["impressions_90d"] > 0]["content_id"].nunique()

# Pages with substantial visibility but low CTR
# Using the guide's starter threshold: impressions >= 500 and CTR < 0.5%
low_ctr_pages = df[
    (df["impressions_90d"] >= 500) &
    (df["ctr"] < 0.5)
]["content_id"].nunique()

print(f"Total unique pages: {total_pages:,}")
print(f"Pages with search impressions: {visible_pages:,}")
print(f"Pages with >=500 impressions and CTR <0.5%: {low_ctr_pages:,}")
```
